#### 3.3.3 Chuẩn hóa danh sách technical_skills

Cột `technical_skills` đã được chuyển thành list ở mục 3.1. Ở đây, chuẩn hóa cách viết bằng alias, gộp phần tử trùng trong từng tin và thống kê số tin chứa từng mục. Các mục chưa có alias được giữ lại để có thể rà soát hoặc phân loại ở bước sau; tần suất thấp không tự động làm mục đó bị loại.


##### Bước 1: Chuẩn hóa chuỗi để so khớp

Bỏ khoảng trắng thừa và không phân biệt chữ hoa, chữ thường khi tra cứu. Tên hiển thị chuẩn vẫn được giữ theo dạng dễ đọc.


```python
import re
from collections import Counter

#Tạo khóa so khớp nhất quán, không làm thay đổi tên hiển thị.
def normalize_key(skill):
    if not isinstance(skill, str):
        return ""
    return re.sub(r"\s+", " ", skill).strip().casefold()
```

##### Bước 2: Khai báo các tên đồng nghĩa

Mỗi khóa trong `ALIAS_GROUPS` là tên chuẩn; danh sách đi kèm là những cách viết được quy về tên đó. Cần rà soát từ điển theo dữ liệu thực tế và chỉ gộp các tên thực sự cùng nghĩa.


```python
ALIAS_GROUPS = {
    "JavaScript": ["javascript", "js"],
    "React": ["react", "reactjs", "react.js"],
    "Vue.js": ["vue", "vuejs", "vue.js framework"],
    "Node.js": ["nodejs", "node js"],
    "Next.js": ["nextjs", "next.js"],
    "PostgreSQL": ["postgres", "postgresql"],
    "Google Cloud Platform": ["gcp", "google cloud"],
    "Microsoft Azure": ["azure"],
    "Kubernetes": ["k8s"],
    "Photoshop": ["adobe photoshop", "photoshop"],
    "Illustrator": ["adobe illustrator", "illustrator"],
    "InDesign": ["adobe indesign", "indesign"],
    "FastAPI": ["fastapi", "fast api"],
    "Scikit-learn": ["scikit-learn", "scikit learn"],
    ".NET": [".net", "dot net", ".netcore", ".net core"],
}

# Bao gồm cả tên chuẩn để tên chuẩn và alias được xử lý cùng một cách.
ALIAS_MAPPING = {
    normalize_key(name): canonical
    for canonical, names in ALIAS_GROUPS.items()
    for name in [canonical, *names]
}
```

```python
# In từ điển để rà soát các cặp tên chuẩn và alias.
print(ALIAS_MAPPING)
```

##### Bước 3: Chuẩn hóa alias và thống kê tần suất

Duyệt trực tiếp các list đã parse ở mục 3.1. Tần suất được đếm theo số tin có chứa mục đó và dùng để xem xét các mục hiếm; danh sách này chưa được lọc theo loại kỹ năng.


```python
#Trả về tên chuẩn theo alias; giữ nguyên tên lạ để tiện rà soát
def canonical_name(skill):
    if not isinstance(skill, str) or not skill.strip():
        return ""
    key = normalize_key(skill)
    return ALIAS_MAPPING.get(key, re.sub(r"\s+", " ", skill).strip())


# Đếm mỗi kỹ năng tối đa một lần trong mỗi tin tuyển dụng.
# Dùng key đã chuẩn hóa để các biến thể khác hoa/thường được đếm chung.
skill_counts = Counter()
skill_names = {}

for skills in df["technical_skills"]:
    if not isinstance(skills, list):
        continue

    keys_in_job = set()

    for skill in skills:
        canonical = canonical_name(skill)
        if not canonical:
            continue

        key = normalize_key(canonical)
        keys_in_job.add(key)

        # Giữ lại tên hiển thị đầu tiên gặp được cho mỗi key
        skill_names.setdefault(key, canonical)

    skill_counts.update(keys_in_job)

vocabulary = {
    key: skill_names[key]
    for key in skill_counts
}

print("Số kỹ năng chuẩn:", len(vocabulary))
print("20 kỹ năng ít gặp nhất:")
print([
    (vocabulary[key], count)
    for key, count in sorted(skill_counts.items(), key=lambda item: item[1])[:20]
])
```

```python
print("Số kỹ năng chuẩn:", len(vocabulary))
print("20 mục đầu tiên:", list(vocabulary.values())[:20])
```

##### Bước 4: Chuẩn hóa từng tin tuyển dụng

Mỗi tin được chuyển thành tập hợp tên chuẩn để không đếm lặp một mục trong cùng giao dịch. Các mục không có alias vẫn được giữ theo giá trị đã chuẩn hóa khoảng trắng; có thể phân loại hoặc lọc ở notebook kế tiếp. Kết quả được lưu ở `normalized_skills`, còn cột gốc được giữ nguyên.


```python
def normalize_skills(skills):
    normalized = set()

    for skill in skills:
        if not isinstance(skill, str) or not skill.strip():
            continue

        key = normalize_key(skill)
        canonical = ALIAS_MAPPING.get(key)
        if canonical is not None:
            normalized.add(vocabulary.get(normalize_key(canonical), canonical))
        else:
            normalized.add(vocabulary.get(key, re.sub(r"\s+", " ", skill).strip()))

    return sorted(normalized, key=str.casefold)


df["normalized_skills"] = df["technical_skills"].apply(normalize_skills)
```

### 3.4 Tạo tập giao dịch (transaction)

```python
transactions = df["normalized_skills"].tolist()

transactions[:3]
```

### 3.5 Mã hóa one-hot

```python
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()

te_array = te.fit(transactions).transform(transactions)

transaction_df = pd.DataFrame(
    te_array,
    columns=te.columns_
)
```
